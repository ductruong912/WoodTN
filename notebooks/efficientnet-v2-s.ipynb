# EfficientNetV2-S training

Two-stage transfer learning with class-balanced sampling, Mixup, CutMix, and early stopping.

Run cells in order. Supply your own class-folder dataset; see the root README.
Set `WOODTN_DATA_DIR` and `WOODTN_OUTPUT_DIR` to override local defaults.
Pretrained ImageNet weights are downloaded on the first training run.
Validation metrics guide model selection and are not held-out test results.


```python
import os
import random
from pathlib import Path

import numpy as np
import torch

ROOT = Path.cwd()
if not (ROOT / "classes.txt").is_file():
    ROOT = ROOT.parent
if not (ROOT / "classes.txt").is_file():
    raise RuntimeError("Start Jupyter from the repository root or notebooks directory.")
DATA_ROOT = Path(os.environ.get("WOODTN_DATA_DIR", ROOT / "data_split"))
OUTPUT_ROOT = Path(os.environ.get("WOODTN_OUTPUT_DIR", ROOT / "checkpoints"))
NUM_WORKERS = 0  # Portable default for notebook execution on Windows.
SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
if torch.cuda.is_available():
    torch.cuda.manual_seed_all(SEED)
```

```python
import copy
import math
import time

import torch.nn as nn
import torch.optim as optim
from torch.optim import lr_scheduler
from torchvision import datasets, models, transforms
from tqdm import tqdm
```

```python
SAVE_DIR = OUTPUT_ROOT / "efficientnet_v2_s"
SAVE_DIR.mkdir(parents=True, exist_ok=True)
```

```python
IMG_SIZE = 224

data_transforms = {
    "train": transforms.Compose(
        [
            transforms.RandomResizedCrop(IMG_SIZE, scale=(0.85, 1.0), ratio=(0.9, 1.1)),
            transforms.RandAugment(num_ops=2, magnitude=10),
            transforms.RandomHorizontalFlip(p=0.5),
            transforms.RandomVerticalFlip(p=0.2),
            transforms.ColorJitter(
                brightness=0.2, contrast=0.25, saturation=0.2, hue=0.02
            ),
            transforms.ToTensor(),
            transforms.RandomErasing(p=0.15, scale=(0.02, 0.10), value=0),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
        ]
    ),
    "val": transforms.Compose(
        [
            transforms.Resize(int(IMG_SIZE * 1.15)),
            transforms.CenterCrop(IMG_SIZE),
            transforms.ToTensor(),
            transforms.Normalize([0.485, 0.456, 0.406], [0.229, 0.224, 0.225]),
        ]
    ),
}
```

```python
from collections import Counter

from torch.utils.data import WeightedRandomSampler

train_root = DATA_ROOT / "train"
val_root = DATA_ROOT / "val"

trainset = datasets.ImageFolder(train_root, transform=data_transforms["train"])
valset = datasets.ImageFolder(val_root, transform=data_transforms["val"])
assert trainset.class_to_idx == valset.class_to_idx, (
    "Class order differs across splits."
)

# Class-balanced sampler
train_targets = [y for _, y in trainset.samples]
class_counts = Counter(train_targets)
class_weights = {c: 1.0 / max(1, class_counts[c]) for c in class_counts}
sample_weights = [class_weights[y] for y in train_targets]
sampler = WeightedRandomSampler(
    weights=torch.as_tensor(sample_weights, dtype=torch.double),
    num_samples=len(sample_weights),
    replacement=True,
)

trainloader = torch.utils.data.DataLoader(
    trainset,
    batch_size=24,
    sampler=sampler,
    num_workers=NUM_WORKERS,
    drop_last=False,
    pin_memory=True,
)
valloader = torch.utils.data.DataLoader(
    valset,
    batch_size=24,
    shuffle=False,
    num_workers=NUM_WORKERS,
    drop_last=False,
    pin_memory=True,
)
```

```python
dataset_sizes = {
    "train": len(trainset),
    "val": len(valset),
}
data_loaders = {"train": trainloader, "val": valloader}
```

```python
# Warmup on top of cosine
class WarmupCosine(torch.optim.lr_scheduler._LRScheduler):
    def __init__(self, optimizer, warmup_epochs, max_epochs, last_epoch=-1):
        self.warmup_epochs = warmup_epochs
        self.max_epochs = max_epochs
        super().__init__(optimizer, last_epoch)

    def get_lr(self):
        if self.last_epoch < self.warmup_epochs:
            warmup_factor = float(self.last_epoch + 1) / float(
                max(1, self.warmup_epochs)
            )
            return [base_lr * warmup_factor for base_lr in self.base_lrs]
        progress = float(self.last_epoch - self.warmup_epochs) / float(
            max(1, self.max_epochs - self.warmup_epochs)
        )
        return [
            0.5 * base_lr * (1.0 + math.cos(math.pi * progress))
            for base_lr in self.base_lrs
        ]
```

```python
device = torch.device(
    "cuda:0"
    if torch.cuda.is_available()
    else ("mps" if torch.backends.mps.is_available() else "cpu")
)
model = models.efficientnet_v2_s(weights=models.EfficientNet_V2_S_Weights.IMAGENET1K_V1)
model.classifier = nn.Sequential(
    nn.Dropout(p=0.3), nn.Linear(in_features=1280, out_features=len(trainset.classes))
)
# Freeze backbone for warmup
for param in model.features.parameters():
    param.requires_grad = False
for param in model.classifier.parameters():
    param.requires_grad = True
model = model.to(device, memory_format=torch.channels_last)
use_amp = device.type == "cuda"
scaler = torch.cuda.amp.GradScaler(enabled=use_amp)
print("Device:", device, "AMP:", use_amp)
```

```python
def mixup_data(x, y, alpha=0.4):
    lam = torch.distributions.Beta(alpha, alpha).sample().item() if alpha > 0 else 1
    index = torch.randperm(x.size(0), device=x.device)
    mixed_x = lam * x + (1 - lam) * x[index]
    y_a, y_b = y, y[index]
    return mixed_x, y_a, y_b, lam


def cutmix_data(x, y, alpha=1.0):
    lam = torch.distributions.Beta(alpha, alpha).sample().item()
    b, _, H, W = x.size()
    index = torch.randperm(b, device=x.device)
    y_a, y_b = y, y[index]

    rw = int(W * math.sqrt(1 - lam))
    rh = int(H * math.sqrt(1 - lam))
    cx = torch.randint(0, W, (1,), device=x.device).item()
    cy = torch.randint(0, H, (1,), device=x.device).item()
    x1 = max(cx - rw // 2, 0)
    y1 = max(cy - rh // 2, 0)
    x2 = min(cx + rw // 2, W)
    y2 = min(cy + rh // 2, H)
    x[:, :, y1:y2, x1:x2] = x[index, :, y1:y2, x1:x2]
    lam = 1 - ((x2 - x1) * (y2 - y1) / (W * H))
    return x, y_a, y_b, lam


def train_model(
    model,
    criterion,
    optimizer,
    scheduler,
    data_loaders,
    dataset_sizes,
    device,
    num_epochs,
    patience=5,
    use_mixup=True,
    use_cutmix=True,
    mixup_alpha=0.4,
    cutmix_alpha=1.0,
    mix_prob=0.3,
):
    since = time.time()
    best_model_wts = copy.deepcopy(model.state_dict())
    best_val_loss = float("inf")
    best_acc = 0.0
    early_stop_counter = 0

    for epoch in range(num_epochs):
        print(f"\nEpoch {epoch}/{num_epochs - 1}")
        print("-" * 10)

        for phase in ["train", "val"]:
            model.train() if phase == "train" else model.eval()

            running_loss = 0.0
            running_corrects = 0
            loop = tqdm(data_loaders[phase], desc=f"{phase} [{epoch + 1}/{num_epochs}]")

            for inputs, labels in loop:
                inputs, labels = (
                    inputs.to(device, non_blocking=True),
                    labels.to(device, non_blocking=True),
                )
                optimizer.zero_grad(set_to_none=True)

                with torch.set_grad_enabled(phase == "train"):
                    mixed = False
                    if phase == "train" and torch.rand(1).item() < mix_prob:
                        if use_mixup and use_cutmix:
                            if torch.rand(1).item() < 0.5:
                                inputs, targets_a, targets_b, lam = mixup_data(
                                    inputs, labels, alpha=mixup_alpha
                                )
                            else:
                                inputs, targets_a, targets_b, lam = cutmix_data(
                                    inputs, labels, alpha=cutmix_alpha
                                )
                            mixed = True
                        elif use_mixup:
                            inputs, targets_a, targets_b, lam = mixup_data(
                                inputs, labels, alpha=mixup_alpha
                            )
                            mixed = True
                        elif use_cutmix:
                            inputs, targets_a, targets_b, lam = cutmix_data(
                                inputs, labels, alpha=cutmix_alpha
                            )
                            mixed = True

                    if use_amp and device.type == "cuda":
                        with torch.cuda.amp.autocast():
                            outputs = model(inputs)
                            if mixed:
                                loss = lam * criterion(outputs, targets_a) + (
                                    1 - lam
                                ) * criterion(outputs, targets_b)
                                acc_targets = targets_a
                            else:
                                loss = criterion(outputs, labels)
                                acc_targets = labels
                        if phase == "train":
                            scaler.scale(loss).backward()
                            scaler.step(optimizer)
                            scaler.update()
                    else:
                        outputs = model(inputs)
                        if mixed:
                            loss = lam * criterion(outputs, targets_a) + (
                                1 - lam
                            ) * criterion(outputs, targets_b)
                            acc_targets = targets_a
                        else:
                            loss = criterion(outputs, labels)
                            acc_targets = labels
                        if phase == "train":
                            loss.backward()
                            optimizer.step()

                running_loss += loss.item() * inputs.size(0)
                _, preds = torch.max(outputs, 1)
                running_corrects += torch.sum(preds == acc_targets)

            if phase == "train" and scheduler is not None:
                scheduler.step()

            epoch_loss = running_loss / dataset_sizes[phase]
            epoch_acc = running_corrects.double() / dataset_sizes[phase]

            print(f"{phase} Loss: {epoch_loss:.4f} Acc: {epoch_acc:.4f}")

            if phase == "val":
                if epoch_loss < best_val_loss:
                    best_val_loss = epoch_loss
                    best_acc = epoch_acc
                    best_model_wts = copy.deepcopy(model.state_dict())
                    early_stop_counter = 0
                    filename_model = SAVE_DIR / "best.pth"
                    torch.save(
                        {
                            "model_state": best_model_wts,
                            "classes": trainset.classes,
                            "epoch": epoch,
                            "val_loss": epoch_loss,
                        },
                        filename_model,
                    )
                    print(f"✅ Model saved: {filename_model}")
                else:
                    early_stop_counter += 1
                    print(
                        f"⏳ No improvement. Early stop counter: {early_stop_counter}/{patience}"
                    )

        if early_stop_counter >= patience:
            print(f"⛔ Early stopping at epoch {epoch}")
            break

    time_elapsed = time.time() - since
    print(f"\nTraining complete in {time_elapsed // 60:.0f}m {time_elapsed % 60:.0f}s")
    print(f"Validation accuracy at lowest loss: {best_acc:.4f}")
    model.load_state_dict(best_model_wts)
    return model
```

```python
import torch.nn as nn

criterion = nn.CrossEntropyLoss(label_smoothing=0.1)
optimizer_ft = optim.AdamW(model.parameters(), lr=3e-4, weight_decay=5e-2)
exp_lr_scheduler = WarmupCosine(optimizer_ft, warmup_epochs=5, max_epochs=10)
```

```python
model_ft = train_model(
    model,
    criterion,
    optimizer_ft,
    exp_lr_scheduler,
    data_loaders,
    dataset_sizes,
    device,
    num_epochs=10,
    patience=3,
    use_mixup=True,
    mixup_alpha=0.4,
    cutmix_alpha=1.0,
    mix_prob=0.3,
)
```

```python
model_ft = train_model(
    model_ft,
    criterion,
    optimizer_ft,
    exp_lr_scheduler,
    data_loaders,
    dataset_sizes,
    device,
    num_epochs=30,
    patience=7,
    use_mixup=True,
    mixup_alpha=0.4,
    cutmix_alpha=1.0,
    mix_prob=0.3,
)
```
